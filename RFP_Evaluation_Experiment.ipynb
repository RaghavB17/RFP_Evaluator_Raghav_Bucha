## 1. Problem Statement
Evaluate multiple proposal PDFs consistently using five weighted criteria: Technical Capability, Implementation Plan, Commercial Value, Security & Compliance, and Support & Experience.

```python
criteria = {
    "Technical Capability": 0.25,
    "Implementation Plan": 0.20,
    "Commercial Value": 0.20,
    "Security & Compliance": 0.20,
    "Support & Experience": 0.15
}
sum(criteria.values())
```

## 2. Prompt Experimentation

### Experiment A — Generic prompt
A generic instruction such as `Evaluate this proposal` can produce inconsistent criteria coverage.

### Experiment B — Structured rubric
The final approach explicitly supplies the five criteria, weights, 1–10 scoring range, evidence-only instruction, and JSON output schema.

### Experiment C — Explainability
The evaluator additionally requests strengths, risks/gaps, and a concise rationale so that scores are auditable.

```python
# Final weighted-score experiment
sample_scores = {
    "Technical Capability": 8,
    "Implementation Plan": 7,
    "Commercial Value": 8,
    "Security & Compliance": 9,
    "Support & Experience": 7
}
weighted = sum(sample_scores[k] * criteria[k] for k in criteria)
absolute_score = weighted * 10
weighted, absolute_score
```

## 3. PPI Experiment

PPI is a relative benchmark against the strongest proposal in the submitted set.

`PPI = proposal score / best proposal score × 100`

```python
scores = [82, 76, 91, 68]
best = max(scores)
ppi = [round(s / best * 100, 2) for s in scores]
ppi
```

## 4. Persistence Design

Evaluation results are stored in SQLite. JSON export is optional for convenience; the database is the primary persisted output.

```python
import sqlite3
con = sqlite3.connect("rfp_evaluation.db")
con.execute('''CREATE TABLE IF NOT EXISTS evaluations (
id INTEGER PRIMARY KEY AUTOINCREMENT,
proposal TEXT, technical REAL, implementation REAL, commercial REAL,
security REAL, support_experience REAL, absolute_score REAL,
ppi REAL, rank INTEGER, rationale TEXT, method TEXT, evaluated_at TEXT)''')
con.commit()
con.close()
print("SQLite schema ready.")
```
